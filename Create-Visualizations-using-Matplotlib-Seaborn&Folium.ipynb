```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset (Standard IBM URL for this project)
URL = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DV0101EN-SkillsNetwork/Data%20Files/historical_automobile_sales.csv"
df = pd.read_csv(URL)
```

```python
# 1.1: Line chart for year-to-year automobile sales
plt.figure(figsize=(10, 6))
df_line = df.groupby('Year')['Automobile_Sales'].mean()
df_line.plot(kind='line')
plt.title('Automobile Sales Fluctuation Over Years')
plt.xlabel('Year')
plt.ylabel('Average Sales')
plt.show()
```

```python
# 1.2: Line chart for advertising vs sales during non-recession
plt.figure(figsize=(10, 6))
non_rec = df[df['Recession'] == 0]
sns.lineplot(data=non_rec, x='Year', y='Automobile_Sales', label='Sales')
sns.lineplot(data=non_rec, x='Year', y='Advertising_Expenditure', label='Advertising')
plt.title('Advertising vs Automobile Sales (Non-Recession)')
plt.show()
```

```python
# 1.3: Bar chart comparing sales per vehicle type (Recession vs Non-Recession)
plt.figure(figsize=(10, 6))
sns.barplot(data=df, x='Recession', y='Automobile_Sales', hue='Vehicle_Type')
plt.title('Sales Trend per Vehicle Type: Recession vs Non-Recession')
plt.xticks(ticks=[0, 1], labels=['Non-Recession', 'Recession'])
plt.show()
```

```python
# 1.4: Subplots for GDP variations
fig = plt.figure(figsize=(12, 6))
ax0 = fig.add_subplot(1, 2, 1)
ax1 = fig.add_subplot(1, 2, 2)
df[df['Recession'] == 1].groupby('Year')['GDP'].mean().plot(kind='line', ax=ax0, title='GDP During Recessions')
df[df['Recession'] == 0].groupby('Year')['GDP'].mean().plot(kind='line', ax=ax1, title='GDP During Non-Recessions')
plt.show()
```

```python
# 1.5: Bubble plot for seasonality impact
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='Month', y='Automobile_Sales', size='Seasonality_Weight', legend=False, sizes=(20, 500))
plt.title('Impact of Seasonality on Automobile Sales')
plt.show()
```

```python
# 1.6: Scatter plot for price vs sales volume during recessions
plt.figure(figsize=(10, 6))
rec_data = df[df['Recession'] == 1]
sns.scatterplot(data=rec_data, x='Price', y='Automobile_Sales')
plt.title('Average Vehicle Price vs Sales Volume (Recession)')
plt.show()
```

```python
# 1.7: Pie chart for advertising expenditure by recession status
ad_exp = df.groupby('Recession')['Advertising_Expenditure'].sum()
ad_exp.plot(kind='pie', labels=['Non-Recession', 'Recession'], autopct='%1.1f%%', figsize=(8, 8))
plt.title('Advertising Expenditure: Recession vs Non-Recession')
plt.ylabel('')
plt.show()
```

```python
# 1.8: Pie chart for ad expenditure per vehicle type (Recession only)
ad_exp_vehicle = rec_data.groupby('Vehicle_Type')['Advertising_Expenditure'].sum()
ad_exp_vehicle.plot(kind='pie', autopct='%1.1f%%', figsize=(8, 8))
plt.title('Advertising Expenditure by Vehicle Type (Recession)')
plt.ylabel('')
plt.show()
```

```python
# 1.9: Line chart for unemployment rate effect on sales by vehicle type
plt.figure(figsize=(10, 6))
sns.lineplot(data=rec_data, x='unemployment_rate', y='Automobile_Sales', hue='Vehicle_Type')
plt.title('Effect of Unemployment Rate on Sales by Vehicle Type (Recession)')
plt.show()
```
